# 6. Predictive modeling

Regression predicts a numeric target. The target here is the next year's reported state/UT accident count. We hold out 2024 chronologically; no random split is used. MAE is a typical absolute miss in accident counts, RMSE penalizes larger misses, and R² compares fit to a mean baseline. The single test year and small panel make conclusions provisional.
