# 4. Time-series analysis

Chronological ordering matters: a lag for year *t* must come from year *t-1*. A rolling mean smooths short-term variation but does not create more independent years. This report gives five India years; the WHO file gives only 2021.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
import data_collection, data_cleaning
data_collection.main()
national, states = data_cleaning.clean_india()
national = national.sort_values('year')
national.assign(accident_yoy_pct=national.accidents.pct_change().mul(100),
               accident_rolling_3yr=national.accidents.rolling(3, min_periods=1).mean())
```

```python
import plotly.express as px
px.line(national, x='year', y=['accidents','fatalities','persons_injured'], markers=True, title='Reported India totals, 2020-2024')
```
