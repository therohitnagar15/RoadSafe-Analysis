# 3. Exploratory data analysis

A histogram describes a distribution; a grouped summary answers a comparison question. State counts represent burden and are not exposure-adjusted risk. Correlation/association does not show causation.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
import data_cleaning
national, states = data_cleaning.clean_india()
who = data_cleaning.clean_global()
national.describe()
```

```python
states.groupby('year')['accidents'].agg(['count','mean','median','sum']).round(1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(data=states.query('year == 2024'), x='accidents', bins=12)
plt.title('State/UT reported accident counts, 2024')
plt.xlabel('Reported accidents')
plt.show()
```

```python
import plotly.express as px
px.bar(states.query('year == 2024').nlargest(12, 'accidents'), x='accidents', y='state_ut', orientation='h', title='12 highest state/UT accident counts, 2024')
```
